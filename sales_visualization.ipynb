```python
import pandas as pd

df = pd.read_csv(
    "../datasets/superstore/extracted/Sample - Superstore.csv",
    encoding="latin1"
)

df.head()
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt

sales_by_category = df.groupby("Category")["Sales"].sum()

sales_by_category.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
profit_by_category = df.groupby("Category")["Profit"].sum()

profit_by_category.plot(kind="bar")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

```python
sales_by_region = df.groupby("Region")["Sales"].sum()

sales_by_region.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
df["Segment"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Customer Segment Distribution")
plt.ylabel("")

plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind="line")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
df.plot(
    kind="scatter",
    x="Sales",
    y="Profit"
)

plt.title("Sales vs Profit")

plt.show()
```

```python
df["Profit"].hist(bins=30)

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products.plot(kind="bar")

plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

## Key Findings

1. Technology category generates the highest sales.

3. Regional sales performance differs across markets.
4. Consumer segment contributes the largest share of customers.
